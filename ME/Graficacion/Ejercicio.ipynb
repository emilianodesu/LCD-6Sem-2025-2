```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
x = np.arange(25)
y = np.random.randint(10, 20, 25)
y[24] = y[0]
fig = plt.figure(figsize=(7, 5))
ax = plt.subplot(111)
ax.fill_between(x, y)
ax.set_xlabel('Hora de posteo (US EST)')
ax.set_ylabel('Porcentaje de envios para primera plana')
ax.set_title('Posibilidad de llegar a primera plana')
ax.grid('on')
```

```python
sns.set_theme(style="darkgrid")

fig, ax = plt.subplots(figsize=(8, 5))

sns.lineplot(x=x, y=y, marker="o", markersize=8, color="royalblue", linewidth=2, ax=ax)
ax.fill_between(x, y, interpolate=True, color="lightsteelblue", alpha=0.5)

max_idx = np.argmax(y)
ax.scatter(x[max_idx], y[max_idx], color="red", s=100, zorder=3, label="Máximo", edgecolor="black")

ax.set_xlabel('Hora de posteo (US EST)', fontsize=12, fontweight='bold')
ax.set_ylabel('Porcentaje de envíos para primera plana', fontsize=12, fontweight='bold')
ax.set_title('Posibilidad de llegar a primera plana', fontsize=14, fontweight='bold')

ax.set_xticks(np.arange(0, 25, 3))
ax.set_yticks(np.arange(min(y), max(y)+1, 1))
ax.set_ylim(min(y) -2, max(y) + 2)

ax.legend()
ax.grid('on')
plt.show()
```
