## Pruebas de hipótesis/pruebas de significancia
En resumidas cuentas **significancia estadística** es la **forma en que l@s estadistas miden si un experimento (o estudio) provee un resultado _más extremo_ de lo que pura aleatoreidad produciría**. Si el resultado está más allá del campo de variación aleatoria, se dice que es _estadísticamente significativo_.

El motivo por el que estamos aquí es usar datos, en este caso para hacer/verificar predicciones. En el caso de pruebas de hipótesis las predicciones, que usualmente resultan de estudios previos o de la teoría, se conocen como _hipótesis_ sobre la población de estudio.

> ### Hipótesis
>
> En estadística, una _hipótesis_ es un enunciado sobre una distribución poblacional. Toma la forma de una predicción sobre la distirbución, como que un parámetro en particular tiene el mismo valor para dos grupos a comparación.

Ejemplos:
- Para todos los empleados de una cadena de comida rápida, el salario promedio es el mismo para mujeres que para hombres.
- Solo la mitad de los adultos en nuestro país están satisfechos con los servicios de salud públicos.

Las hipótesis son investigadas con referencia a _hipótesis alternativas_, como "Para todos los empleados de una cadena de comida rápida, el salario promedio es _mayor_ para hombres que para mujeres".

Una **_prueba de significancia_** estadística utiliza datos para resumir la evidencia sobre una hipótesis al comparar una estimación puntual del parámetro de interés al valor predicho por la hipótesis.

### Ejemplo: Prueba para el sesgo en la selección de mánagers.

Una cadena de supermercados en el sureste de USA selecciona empleados periódicamente para recibir capacitacipon en gerencia. Un grupo de empleadas asegura que la compañia selecciona desproporcionadamente a más hombres para esta capacitación (http://now.org/blog/walmart-and-sex-discrimination).

Supongamos que el pool de empleados a consideración para la capacitación gerencial es grande y consiste de un 40% de hombres y 60% de mujeres.

- La aseveración por parte del supermercado de que no hay tal sesgo de género se traduce en la hipótesis de que (siendo lo demás igual, e.g., selección aleatoria, sin importar edad o antiguedad), en cada selección la probabilidad de elegir un hombre es igual a 0.40 y la probabilidad de elegir a una mujer es 0.60.
- La declaración de las empleadas es una hipótesis alternativa de que la probabilidad de escoger a un hombre es mayor a 0.40.

Supongamos ahora que en la última selección de empleados para capacitación, 9 de los 10 escogidos fueron hombres. Aún siendo seleccionados aleatoriamente del pool de empleados, debido a variación muestral puede que sea posible que 9 hombres de 10 escogidos sea posible.

Para evaluar lo anterior, debemos analizar si esta selección resultante **sería poco probable** (unlikely), si **no** hubiera discriminación por razón de género.

#### Elementos de una prueba de hipótesis
1. Suposiciones

   Como muchos métodos estadísticos, cada prueba hace ciertas suposiciones o necesita que ciertas condiciones sean válidas. Estas se relacionan con: tipos de datos, aleatorización, distribución poblacional, tamaño de muestra, entre otros.

2. Hipótesis

    Las pruebas de significancia/de hipótesis tienen dos hipótesis acerca de las distribuciones poblacionales, usualmente expresadas en términos de parámetros de estas distribuciones.

   > #### Hipótesis nula e hipótesis alterna
   >
   > La _hipótesis nula_, denotada por $H_0$, es una afirmación de que un parámetro toma un valor particular. La _hipótesis alterna_, denotada por $H_a$, enuncia que un parámetro toma valores en un rango alternativo. Usualmente el valor en $H_0$ corresponde, en un cierto sentido, a _efecto nulo_. Entonces, los valores en $H_a$ representan un tipo particular de efecto.
   >
    **Las hipótesis son formuladas antes de analizar los datos**. La hipótesis alternativa es usualmente una hipótesis de investigación que se cree cierta.

   En el ejemplo a consideración, sea $\pi$ la probabilidad de que cualquier selección particular sea hombre. La afirmación de la compañía de que no hay sesgo significa: $\pi=0.40$, el cual es un ejemplo de hipótesis nula, _efecto nulo_, refiriéndose a una falta de sesgo de género. La hipótesis alternativa refleja la creencia de las empleadas de que esta probabilidad en realidad sobrepasa $0.40$. En resumen, $H_0: \pi=0.40$ y $H_a:\pi >40$. Aún cuando $H_0$ tiene un solo valor, $H_a$ tiene un rango de valores.
4.  Estadístico de prueba (Test statistic)

    El parámetro al que se refiere la hipótesis es una estimación puntual. El _estadístico de prueba_ resume qué tan lejos esta estimación puntual cae del valor del parámetro en $H_0$.


5. P-valor y conclusión


   Diferentes pruebas utilizan diferentes estadísticos, interpretaciones más accesibles resultan de tener un resumen _probabilístico_ de la evidencia en contra de $H_0$. Para esto, utilizamos la distribución muestral del estadístico de prueba, **bajo la suposición de que $H_0$ es verdad**.

   > #### P-valor
   >
   > El P-valor es la probabilidad, suponiendo que $H_0$ es verdad, de que el estadístico de prueba sea igual al valor observado o a un valor aún más extremo en la dirección predicha por $H_a$.
   >

Un P-valor pequeño, como $0.01$, significa que los datos observados serían inusuales si $H_0$ fuera verdad. Un valor moderado o grande del P-valor, como $0.26$ o $0.83$, significa que los datos son consistentes con $H_0$. _P-valores pequeños reflejan evidencia fuerte en contra de $H_0$_.

Regresando al ejemplo de los aprendices de gerencia, para probar $H_0:\pi=0.40$ en contra de $H_a:\pi>0.40$, con $n=10$ es suficiente por el momento utilizar como estadístico de prueba el conteo muestral $y=9$, considerándolo como el resultado de una variable aleatoria binomial. Los valores de $y$ que proveen esta cantidad de evidencia extrema (o más) en contra de $H_0:\pi=0.40$ y a favor de $H_a:\pi>0.40$ son los valores de cola-derecha $\ge9$, en palabras, $9$ 0 $10$ hombres en una selección de $10$.

De la distribución binomial tenemos que
$$
    \mathbb{P}[Y\ge9]=\binom{10}{9}(0.4)^2(0.6)+\binom{10}{10}(0.4)^10(0.6)^0=0.0017,
$$
y este es el P-valor. Si las selecciones realmente fueran aleatorias con respecto al género, la probabilidad de un resultado como el observado es tan solo $0.0017$, dando fuerte evidencia en contra de $H_0$ y apoyando $H_0>0.4$.

A veces es necesario realizar una decisión sobre la validez de $H_0$. Si el P-valor es lo suficientemente pequeño, rechazamos $H_0$ y aceptamos $H_a$. Generalmente se sugiere (requiere) un P-valor pequeño, como $0.05$, para rechazar $H_0$. Se dice entonces que los resultados son _significativos_ al nivel $0.05$.

## Recap:
- El propósito de las pruebas de significancia/pruebas de hipótesis es ayudar en identificar si un efecto observado es debido a pura casualidad (random chance) o no.
- Esto es debido a la tendencia que se tiene en subestimar o sobrestimar (dependiendo del mindset) el comportamiento natural aleatorio o malinterpretar que eventos aleatorios tengan patrones o alguna significancia que en realidad no tienen.
- Cuando se presume que un efecto observado no es debido solamente a casualidad se dice que es _estadísticamente significativo_.

## Pruebas de significancia para proporciones y medias
Las pruebas más sencillas son respecto a una proporcion o media para una sola muestra.

### Prueba de significancia para una proporción
Para variables categóricas, los parámetros son las proporciones poblacionales para las categorías. En el caso binario solo tenemos un parámetro: $\pi$.

- Suposiciones

  Suponemos que es una muestra aleatoria y que el tamaño de muestra es suficientemente grande como para que la distr. muestral de
  $$
    \hat{\pi}=\frac{1}{n}\sum_{i=1}^nX_i
  $$
  sea aprox. normal. Para obtener una forma acampanada-simétrica para la distribución muestral se requiere un mayor $n$ cuando $\pi$ está cerca de 0 o 1.
- Hipótesis

    La hipótesis nula tiene la forma
    $$
        H_0:\pi=\pi_0
    $$
    donde $\pi_0$ denota un valor particular entre 0 y 1, como $H_0:\pi=0.5$. La hipótesis alternativa **más común** es
  $$
    H_a:\pi\neq\pi_0,\quad\text{e.g.}\quad\pi\neq0.5\text{.}
  $$
  Esta hip. alterna es llamada _bilateral_ (two-sided).

- Estadístico de prueba

  Tenemos que, **suponiendo $H_0$ como cierta**
  $$
    \hat{\pi}\overset{\text{aprox}}{\sim}\text{N}\left(\pi_0,\ \frac{\pi_0(1-\pi_0)}{n}\right)\text{,}
  $$
  luego, el estadístico de prueba _score_ es
  $$
      Z=\frac{\hat{\pi}-\pi_0}{se_0}\overset{\text{aprox}}{\sim}\text{N}(0,1),\ se_0=\sqrt{\frac{\pi_0(1-\pi_0)}{n}}\text{.}
  $$

- P-valor y conclusión

  Para $H_a:\pi\neq\pi_0$, valores $z$ más extremos son aquellos aún más lejos **en las dos colas** de la distr. normal estándar y el P-valor es la probabilidad **de ambas colas**  de que $|Z|$ sea **al menos** tan grande como el $|z|$ observado:
  $$
  \mathbb{P}\left[|Z|\ge|z|\right]\text{.}
  $$

### Ejemplo
Una encuesta de Fox News entrevistó una muestra aleatoria de 1008 Estadounidenses en 2019, preguntándoles si pensaban que el cambio climático es una amenaza mayor para la estabilidad de Estados Unidos. Sea $\pi$ la proporción poblacional que respondería afirmativamente. Para analizar si es una minoría o mayoría de la población, probamos
$$
    H_0:\pi=0.5\quad\text{vs}\quad H_a:\pi\neq0.5\text{.}
$$
Suponiendo que $H_0$ es verdad

$$
    se_0=\sqrt{\frac{\pi_0(1-\pi_0)}{n}}=\frac{(0.5)(0.5)}{1008}=0.0157\text{.}
$$
En la encuesta 524 del total respondió _sí_. Con $\hat{\pi}=524/1008=0.520$ el estadístico de prueba es
$$
    z = \frac{\hat{\pi}-\pi_0}{se_0}=\frac{0.520-0.50}{0.0157}=1.26.
$$
Luego, $\hat{\pi}$ cae $1.26\times se_0$ por arriba de $\pi_0=0.5$. El P-valor es la prob. de que
$$
    Z\ge1.26\quad\text{o}\quad Z\le-1.26\text{,}
$$
de donde el P-valor bilateral es $2(0.104)=0.208$




- Este valor no es muy pequeño, por lo que la evidencia en contra de $H_0$ no es fuerte.
- Es plausible que $\pi=0.5$.
- No hay evidencia suficiente para determinar si una mayoría o minoría de la población cree que el cambio climático es una amenaza mayor.
- Se debe redondear el P-valor antes de reportarlo, e.g. en este caso a 0.21.
- En este caso la conclusión a veces se reporta como "No rechazar $H_0$", ya que los datos no contradicen $H_0$, por otro lado, no es apropiado decir "Acepta $H_0$".

```python
est_prueba, P_valor = proportions_ztest(524, 1008, 0.5)
print('Valor estadistico de prueba: {0:0.4},'.format(est_prueba), 'P-valor: {0:0.4f}'.format(P_valor))
```

```python
proportion_confint(524, 1008, method = 'wilson')
```

### Prueba de significancia para una media


Para variables cuantitativas, las pruebas de significancia generalmente tienen que ver con **medias poblacionales**.

- Suposiciones

   Se hace la suposición que los datos se obtienen con aleatorización y que estos vienen de una distribución poblacional normal. Esta suposición es importante principalmente para $n$ pequeño e $H_a$ unilaterales. Porque de otra manera es una prueba _robusta_.

- Hipótesis

  La hipótesis nula tiene la forma
    $$
        H_0:\mu=\mu_0\text{,}
    $$

  para un valor particular $\mu_0$. La hipótesis alterna más común es
  $$
    H_a:\mu\neq\mu_0\text{, e.g. } H_a:\mu\neq0.
  $$
    
- Estadístico de prueba

  Recordemos que si $Y_1,...,Y_n\overset{i.i.d.}{\sim}\text{N}(\mu,\sigma)$ entonces $\bar{Y}=\frac{1}{n}\sum_{i=1}^nY_i\sim\text{N}(\mu,\sigma/n)$. Como es usual, tenemos que $\sigma$ es desconocido, lo sustituimos por su versión muestral: $s$ (desv. est. muestral), y con este tenemos el error estándar _estimado_ para nuestro $\bar{Y}$. Así, el estadístico de prueba resulta

  $$
    T=\frac{\bar{Y}-\mu_0}{SE}\text{, con } SE=\frac{S}{\sqrt{n}}\text{.}
  $$

  Su distribución _nula_ es la distribución _t_ con grados de libertad $n-1$.


- P-valor y conclusión

  El P-valor es la prob., **suponiendo $H_0$**, que el estadístico de prueba anterior: $T$ (es una v.a. **antes** de que observemos los datos), sea igual al valor observado $t$ **o un valor más extremo** que dé evidencia aún más fuerte **en contra** de $H_0$. Para $H_a:\mu\neq\mu_0$, el P-valor es la probabilidad de **ambas colas** de que $|T|$ sea al menos tan grande que $|t|$ (el valor que eventualmente observamos al recibir los datos)
    $$
        \mathbb{P}[|T|\ge|t|]\text{.}
    $$


### Ejemplo

En promedio, ¿Qué tan conservadores o liberales son los ciudadanos estadounidenses?. Utilizando una sección del General Social Survey haremos una prueba de significancia para la media poblacional en el caso de la población hispana.


Ideología política es una escala ordinal, pero es en ocasiones informativo tratar datos ordinales para resumir respuestas. En este caso un promedio debajo de 4 muestra una propensión hacia liberalismo y visceversa. Para poner a prueba la pregunta de si los datos dan evidencia para alguna de estas posturas, realizamos la prueba de significancia con hipótesis
$$
    H_0:\mu=4.0\text{,}\quad H_a:\mu\neq4.0\text{.}
$$
Las 369 observaciones hispanas arrojan
$$
    \bar{y}=4.089\text{, }s = 1.339\quad\text{y}\quad se=\frac{s}{\sqrt{n}}=\frac{1.339}{\sqrt{369}}=0.0697\text{,}
$$
por lo que el valor **observado** de $T$ es
$$
    t = \frac{\bar{y}-\mu_0}{se}=\frac{\bar{y}-4.0}{0.0697}=1.283\text{,}
$$
con $df=369-1=368$. Luego, la media muestrael está a una distancia de $1.283\times se$ del valor dado por $H_0$. Así, el P-valor resulta ser $0.20$. Si la ideología media poblacional (todos l@s hispanos en USA) $\mu$ fuera $4.0$ entonces la prob. de que $\bar{Y}$ (para $n=369$, i.e., este caso particular) esté al menos tan lejos de $\mu_0=4.0$ como el valor $\bar{y}=4.089$ es de 0.20. No es inusual. Es plausible que $\mu$ sea $4.0$, no en la dirección liberal o conservadora.


```python
ident_pol = pd.read_csv('http://stat4ds.rwth-aachen.de/data/Polid.dat', sep = r'\s+')
ident_pol.head()
```

```python
# De acuerdo con la doc de stats: calculate the T-test for the mean of ONE group of scores.
stats.ttest_1samp(ident_pol[ident_pol.race == 'hispanic'].ideology, 4.0)    # H_0: \mu_0=4.0
```

```python
# Calculemos un intervalo de confianza del 95% como sabemos hacerlo
u = ident_pol[ident_pol.race == 'hispanic'].ideology.mean()
s = ident_pol[ident_pol.race == 'hispanic'].ideology.std(ddof = 1)
n = ident_pol[ident_pol.race == 'hispanic'].ideology.size
ci_95 = [u-t.isf(0.05/2, n-1)*(s/np.sqrt(n)),
         u+t.isf(0.05/2, n-1)*(s/np.sqrt(n))]
print(ci_95)
```

### Pruebas de significancia comparando media**s**

En la práctica (e.g. A/B testing), inferencias sobre medias o proporciones usualmente involucran comparaciones para diferentes grupos.

Para esta prueba se supone que se tienen dos muestras aleatorias $Y_{11}, ..., Y_{n1}$ y $Y_{12}, ..., Y_{m2}$ que provienen de una misma familia distribucional: la distribución normal. Como en la sección sobre intervalos de confianza, el método usual/estándar supone que las desviaciones estándar poblacionales son iguales
$$
    \sigma_1=\sigma_2\text{.}
$$
Para comparar las medias poblacionales $\mu_1$ y $\mu_2$ se prueba la hipótesis nula $\mu_1=\mu_2\iff\mu_1-\mu_2=0$, en otras palabras
$$
    H_0: Y_{i1}\quad\text{y}\quad Y_{j2}\quad\text{tienen distr.}\quad \text{N}(\mu,\sigma)\quad \text{vs}\quad H_a:Y_{i1}\sim \text{N}(\mu_1,\sigma)\quad\text{y}\quad Y_{j2}\sim\text{N}(\mu_2,\sigma),\quad\mu_1\neq\mu_2\text{.}
$$
Como en la sección de intervalos de confianza **x2** tenemos que estimamos el valor común de $\sigma$ con el pooled estimate $s$.

Este modelo de hipótesis alternativa bilateral es robusto ante violaciones de los supuestos, especialmente cuando los tamaños de muestra son similares y no muy pequeños.

Para probar $H_0:\mu_1=\mu_2$ el estadístico de prueba es
$$
    T=\frac{(\bar{Y}_1-\bar{Y}_2)-0}{SE}
$$
con el error estándar estimado por
$$
    se=\sqrt{\frac{s^2}{n_1}+\frac{s^2}{n_2}}=s\sqrt{\frac{1}{n_1}+\frac{1}{n_2}}
$$
con $s$ el pooled estimate.

Bajo $H_0$ el estadístico $T$ tiene una distr. $t$ con $df=n_1+n_2-2$.

Si los datos muestran evidencia (por ejemplo, al ver sus histogramas) de desv. estándar muy diferentes, es mejor usar en este caso una prueba alternativa conocida como _prueba t de Welch_, donde no se supone $\sigma_1=\sigma_2$. Usa el mismo estadístico de prueba $T$ pero ahora con
$$
    se=\sqrt{\frac{s_1^2}{n_1}+\frac{s_2^2}{n_2}}
$$
y una expresión más elaborada para los grados de libertad.

Sin embargo, cuando sea aproximadamente válido se prefiere la prueba que presupone $\sigma_1=\sigma_2$ ya que en estudios que emplean aleatorización la hipótesis nula ("efecto nulo/sin efecto") corresponde con que las distribuciones sean completamente iguales, no solo las medias.

### Ejemplo: Comparar terapia a grupo de control

En la parte de intervalos de confianza se analizaron los resultados de tratar a mujeres jóvenes con anorexia a través de una terapia cognitivo conductual, comparando el cambio promedio en su peso con el observado en un grupo de control. Ahora probamos
$$
    H_0:\mu_1=\mu_2\quad\text{vs}\quad H_a:\mu_1\neq\mu_2\text{.}
$$

```python
datos = pd.read_csv('https://stat4ds.rwth-aachen.de/data/Anorexia.dat', sep = r"\s+", index_col = 0)
datos['diferencia'] = datos['after']-datos['before']
```

```python
cb = datos[datos['therapy']=='cb']['diferencia']
control = datos[datos['therapy']=='c']['diferencia']
```

```python
help(stats.ttest_ind)
```

```python
res = stats.ttest_ind(cb, control)    # Por default considera varianzas iguales
print('Valor T:', res.statistic)
print('P-valor:', res.pvalue)
print('Grados de libertad:', res.df)
```

## Pruebas no paramétricas
Como vimos, para comparar dos medias la prueba anterior supone que las distribuciones poblacionales son normales, con la misma varianza. Aún cuando el caso con $H_a$ bilateral es robusto, en algunos casos podríamos preferir comparar grupos sin esta suposición.

La siguiente prueba es un ejemplo de _métodos estadísticos no paramétricos_, los cuales, como han visto, son métodos de inferencia que no presuponen una familia distribucional particular para la población. Para estas pruebas, $H_0$ expresa que las distribuciones poblacionales son idénticas para ambos grupos, sin especificar su forma.

Además, estas pruebas vienen con las siguientes suposiciones

- Observaciones independientes de los dos grupos
- Obtenidas con aleatorización

i.e., una muestra aleatoria para cada grupo.
### Prueba de permutación para comparar dos grupos

Una prueba de permutación deriva la distr. muestral de un estadístico (por ej. la diferencia entre medias muestrales) al evaluar todas las permutaciones posibles de cómo las observaciones totales ($n_1+n_2$) pudieron ser colocadas a cada grupo, con $n_1$ en el primero y $n_2$ en el segundo. El primer paso en esta prueba es combinar las observaciones. Esta es la manifestación lógica de la hipótesis nula de que ambas observaciones son equivalentes. Bajo $H_0$ de distribuciones poblacionales idénticas, todas las permutaciones son equiprobables. El P-valor es la proporción de estas para las cuales el estadístico escogido es al menos tan extremo como el observado.  

Al incrementarse $n_1$ y $n_2$, se vuelve más demandante computacionalmente evaluar todas las posibles permutaciones, por lo que el enfoque alternativo es una muestra aleatoria grande (e.g. 1M) de estas.

The permutation test can be computationally expensive and not necessarily
    more accurate than the analytical test, but it does not make strong
    assumptions about the shape of the underlying distribution.
