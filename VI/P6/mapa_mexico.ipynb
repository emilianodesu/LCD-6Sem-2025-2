```python
# Create a Folium map centered on Mexico
mexico_map = folium.Map(location=[23.6345, -102.5528], zoom_start=5, tiles="cartodb positron")  # Center of Mexico

# Add markers for each state capital
for index, row in df.iterrows():
    folium.Marker(
        location=[row['Latitude'], row['Longitude']],
        popup=row['Capital'],  # Show capital name in popup
        tooltip=row['Capital'],  # Show capital name on hover
        icon=folium.Icon(color='blue', icon='info-sign')  # Customize marker icon
    ).add_to(mexico_map)

# Save the map to an HTML file (optional)
# mexico_map.save('mexico_capitals_map.html')

# Display the map in a Jupyter Notebook or browser
mexico_map
```
